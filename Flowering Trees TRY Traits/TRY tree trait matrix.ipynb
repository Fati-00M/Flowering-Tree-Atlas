```python
import pandas as pd

file_path = "try_data/48214.txt"

# Load tree species and create a set for efficient lookup
tree_species_df = pd.read_csv("tree_species.csv")
tree_set = set(tree_species_df["SpeciesName"].astype(str).str.strip())

results = []

for chunk in pd.read_csv(
    file_path,
    sep="\t",
    encoding="latin1",
    chunksize=100000,
    low_memory=False
):

    chunk["SpeciesName"] = chunk["SpeciesName"].astype(str).str.strip()
    chunk["TraitName"] = chunk["TraitName"].astype(str).str.strip()

    # keep only tree species
    chunk = chunk[chunk["SpeciesName"].isin(tree_set)]

    # keep only flowering trait
    flowering = chunk[
        (chunk["TraitName"] == "Plant flowering")
    ][["SpeciesName", "OrigValueStr"]]

    results.append(flowering)

flowering_trees = pd.concat(results, ignore_index=True)
```

```python
import pandas as pd

tree_species = pd.read_csv("tree_species.csv")
tree_set = set(tree_species["SpeciesName"].astype(str).str.strip())
```

```python
import pandas as pd

file_path = "try_data/48214.txt"

chunks = []

for chunk in pd.read_csv(
    file_path,
    sep="\t",
    encoding="latin1",
    chunksize=100000,
    low_memory=False
):

    chunk["SpeciesName"] = chunk["SpeciesName"].astype(str).str.strip()

    # keep only trees
    chunk = chunk[chunk["SpeciesName"].isin(tree_set)]

    # keep only traits you care about
    traits = [
        "Plant height",
        "Plant woodiness",
        "Plant growth form",
        "Plant lifespan (longevity)",
        "Fruit type",
        "Seed germination lag time",
        "Flower color",
        "Flower sex",
        "Wood growth ring distinction",
        "Seed germination temperature",
        "Seed germination requirement",
        "Flowering requirement (requirement for fertility)",
        "Plant human usage types",
        "Seed germination type",
        "Flowering lag time: number of days from exposition start to flowering",
        "Plant flowering",
        "Seed germination base temperature",
        "Seed germination base water potential",
        "Plant growth temperature range",
        "Plant lifespan: age trees reach in forested stands"
    ]

    chunk = chunk[chunk["TraitName"].isin(traits)]

    # keep useful columns
    chunk = chunk[["SpeciesName", "TraitName", "StdValue", "StdValueStr", "OrigValueStr"]]

    chunks.append(chunk)

df = pd.concat(chunks, ignore_index=True)
```

```python
df["Value"] = (
    df["StdValue"]
    .combine_first(df["StdValueStr"])
    .combine_first(df["OrigValueStr"])
)
```

```python
trait_matrix = df.pivot_table(
    index="SpeciesName",
    columns="TraitName",
    values="Value",
    aggfunc="first"   # important: avoids duplicates crashing
)
```

```python
trait_matrix = trait_matrix.reset_index()

print(trait_matrix.shape)
trait_matrix.head()
```

```python
trait_matrix["Plant height"] = pd.to_numeric(trait_matrix["Plant height"], errors="coerce")
trait_matrix["Plant flowering"] = pd.to_numeric(trait_matrix["Plant flowering"], errors="coerce")
```

```python
trait_matrix.to_csv("tree_trait_matrix.csv", index=False)
```
